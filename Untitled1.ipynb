```python
# Google PageRank Mini (B2)
# Mathematical concept: Eigenvalue = 1, eigenvector = steady state,
# Power Iteration / Power Method

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Make NumPy output easier to read
np.set_printoptions(precision=6, suppress=True)
```

```python
# ============================================================
# USER CONFIGURATION
# ============================================================

# Names of the 5 web pages
pages = ["Page A", "Page B", "Page C", "Page D", "Page E"]

# Links between pages
# Example:
# Page A -> Page B means A contains a link to B

links = {
    "Page A": ["Page B", "Page C"],
    "Page B": ["Page C", "Page D"],
    "Page C": ["Page A"],
    "Page D": ["Page C", "Page E"],
    "Page E": ["Page A", "Page D"]
}

# PageRank damping factor
# Standard PageRank commonly uses d = 0.85
damping = 0.85

# Power iteration settings
max_iterations = 100
tolerance = 1e-10

print("Pages:", pages)
print("Damping factor:", damping)
print("Maximum iterations:", max_iterations)
print("Tolerance:", tolerance)
```

```python
# ============================================================
# MATHEMATICAL ASSUMPTIONS
# ============================================================

print("""
ASSUMPTIONS USED:

1. We have exactly 5 web pages.
2. Every outgoing link from a page has equal probability.
3. If a page has no outgoing links (a dangling page),
   its probability is distributed equally among all pages.
4. Damping factor d = 0.85.
5. With probability d, the surfer follows a link.
6. With probability (1-d), the surfer jumps to any page
   with equal probability.
7. PageRank is normalized so that all PageRank values sum to 1.

Mathematical model:

    r = d P r + (1-d)v

where:
    r = PageRank vector
    P = transition probability matrix
    d = damping factor
    v = teleportation vector

At convergence:

    r = G r

Therefore:

    G r = 1 r

So the PageRank vector is an eigenvector of G
corresponding to eigenvalue 1.
""")
```

```python
# ============================================================
# BUILD THE LINK TRANSITION MATRIX
# ============================================================

n = len(pages)

# Create an n x n matrix filled with zeros
P = np.zeros((n, n))

# Build transition probabilities
for i, page in enumerate(pages):

    outgoing = links.get(page, [])

    # Handle dangling pages
    if len(outgoing) == 0:
        P[:, i] = 1 / n

    else:
        probability = 1 / len(outgoing)

        for destination in outgoing:
            j = pages.index(destination)
            P[j, i] = probability

print("Link Transition Matrix P:")
print(P)

print("\nColumn sums:")
print(P.sum(axis=0))
```

```python
# ============================================================
# READABLE MATRIX DISPLAY
# ============================================================

P_table = pd.DataFrame(
    P,
    index=pages,
    columns=pages
)

print("Transition Probability Matrix P:")
display(P_table.style.format("{:.3f}"))
```

```python
# ============================================================
# GOOGLE MATRIX
# ============================================================

# Equal-probability teleportation vector
v = np.ones(n) / n

# Google matrix:
#
# G = dP + (1-d) * v * 1^T
#
# The outer product creates the teleportation matrix.

G = damping * P + (1 - damping) * np.outer(v, np.ones(n))

print("Google Matrix G:")
print(G)

print("\nColumn sums:")
print(G.sum(axis=0))
```

```python
# ============================================================
# POWER ITERATION
# ============================================================

# Start with equal PageRank for every page
rank = np.ones(n) / n

history = [rank.copy()]

for iteration in range(1, max_iterations + 1):

    # Main PageRank iteration:
    #
    # r_(k+1) = G r_k
    #
    new_rank = G @ rank

    # Measure how much the vector changed
    error = np.linalg.norm(new_rank - rank, ord=1)

    rank = new_rank
    history.append(rank.copy())

    # Stop when the solution has converged
    if error < tolerance:
        break

history = np.array(history)

print("Power iteration completed.")
print("Iterations:", iteration)
print("Final error:", error)
print("Final PageRank vector:")
print(rank)

print("\nSum of PageRank values:", rank.sum())
```

```python
# ============================================================
# PAGE RANK RESULTS
# ============================================================

results = pd.DataFrame({
    "Page": pages,
    "PageRank": rank
})

# Add percentage
results["Percentage"] = results["PageRank"] * 100

# Sort from highest to lowest
results = results.sort_values(
    by="PageRank",
    ascending=False
).reset_index(drop=True)

print("FINAL PAGERANK RESULTS")
display(
    results.style
    .format({
        "PageRank": "{:.6f}",
        "Percentage": "{:.2f}%"
    })
)

most_important = results.iloc[0]["Page"]

print(f"\nMost important page: {most_important}")
print(f"PageRank: {results.iloc[0]['PageRank']:.6f}")
```

```python
# ============================================================
# EIGENVALUE = 1 VERIFICATION
# ============================================================

# At convergence:
#
# G r = r
#
# Therefore:
#
# G r - r = 0
#
# This means r is an eigenvector of G
# corresponding to eigenvalue 1.

eigen_result = G @ rank

eigen_error = np.linalg.norm(eigen_result - rank)

print("G × PageRank:")
print(eigen_result)

print("\nPageRank:")
print(rank)

print("\nEigenvector verification error:")
print(eigen_error)

if eigen_error < tolerance:
    print("\n✓ PageRank satisfies G × r ≈ r")
    print("✓ Therefore eigenvalue λ ≈ 1")
else:
    print("\n✗ Further iterations may be required.")
```

```python
# ============================================================
# DIRECT EIGENVALUE CHECK
# ============================================================

eigenvalues, eigenvectors = np.linalg.eig(G)

print("Eigenvalues of the Google Matrix:")

for i, value in enumerate(eigenvalues):
    print(f"Eigenvalue {i+1}: {value}")

# Find eigenvalue closest to 1
closest_index = np.argmin(np.abs(eigenvalues - 1))
closest_eigenvalue = eigenvalues[closest_index]

print("\nEigenvalue closest to 1:")
print(closest_eigenvalue)
```

```python
# ============================================================
# BENCHMARK TEST
# ============================================================

def pagerank_power_iteration(P, damping=0.85,
                             max_iterations=100,
                             tolerance=1e-10):

    n = P.shape[0]

    # Equal teleportation probability
    v = np.ones(n) / n

    # Google matrix
    G = damping * P + (1 - damping) * np.outer(v, np.ones(n))

    # Initial rank
    rank = np.ones(n) / n

    for iteration in range(1, max_iterations + 1):

        new_rank = G @ rank

        error = np.linalg.norm(new_rank - rank, ord=1)

        rank = new_rank

        if error < tolerance:
            break

    return rank, G, iteration, error


# ------------------------------------------------------------
# Benchmark dataset
# ------------------------------------------------------------

benchmark_pages = [
    "A", "B", "C", "D", "E"
]

benchmark_links = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A"],
    "D": ["C", "E"],
    "E": ["A", "D"]
}

# Build benchmark transition matrix
N = len(benchmark_pages)

benchmark_P = np.zeros((N, N))

for i, page in enumerate(benchmark_pages):

    outgoing = benchmark_links[page]

    for destination in outgoing:

        j = benchmark_pages.index(destination)

        benchmark_P[j, i] = 1 / len(outgoing)


# Run PageRank
benchmark_rank, benchmark_G, benchmark_iterations, benchmark_error = \
    pagerank_power_iteration(benchmark_P)


# Display benchmark
benchmark_results = pd.DataFrame({
    "Page": benchmark_pages,
    "PageRank": benchmark_rank
})

benchmark_results["Percentage"] = benchmark_results["PageRank"] * 100

benchmark_results = benchmark_results.sort_values(
    "PageRank",
    ascending=False
).reset_index(drop=True)

print("BENCHMARK RESULTS")
display(
    benchmark_results.style.format({
        "PageRank": "{:.6f}",
        "Percentage": "{:.2f}%"
    })
)

print("Iterations:", benchmark_iterations)
print("Final error:", benchmark_error)
print(
    "Benchmark test passed:",
    abs(benchmark_rank.sum() - 1) < 1e-10
)
```

```python
# ============================================================
# GOOGLE PAGERANK MINI (B2)
# B.Tech Mathematics Mini Project
# ============================================================
#
# Mathematical concepts:
#   1. Transition Probability Matrix
#   2. Google Matrix
#   3. Eigenvalue = 1
#   4. Eigenvector = Steady State
#   5. Power Iteration / Power Method
#
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. INPUT CONFIGURATION
# ------------------------------------------------------------

pages = ["A", "B", "C", "D", "E"]

# Web-page links
links = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A"],
    "D": ["C", "E"],
    "E": ["A", "D"]
}

# Standard PageRank damping factor
damping = 0.85

max_iterations = 100
tolerance = 1e-10


# ------------------------------------------------------------
# 2. DISPLAY PROJECT INFORMATION
# ------------------------------------------------------------

print("=" * 65)
print("                 GOOGLE PAGERANK MINI (B2)")
print("=" * 65)

print("\nPages:")
print(pages)

print("\nWeb Links:")
for page, destinations in links.items():
    print(f"Page {page} -> {', '.join(destinations)}")

print("\nDamping factor:", damping)


# ------------------------------------------------------------
# 3. BUILD TRANSITION MATRIX P
# ------------------------------------------------------------

n = len(pages)

P = np.zeros((n, n))

for i, source in enumerate(pages):

    destinations = links.get(source, [])

    # Dangling page handling
    if len(destinations) == 0:
        P[:, i] = 1 / n

    else:
        probability = 1 / len(destinations)

        for destination in destinations:

            j = pages.index(destination)

            P[j, i] = probability


print("\n" + "=" * 65)
print("TRANSITION PROBABILITY MATRIX P")
print("=" * 65)

P_table = pd.DataFrame(
    P,
    index=[f"Page {p}" for p in pages],
    columns=[f"Page {p}" for p in pages]
)

display(P_table.style.format("{:.3f}"))


# ------------------------------------------------------------
# 4. CHECK TRANSITION MATRIX
# ------------------------------------------------------------

column_sums = P.sum(axis=0)

print("\nColumn sums:")
print(column_sums)

if np.allclose(column_sums, 1):
    print("✓ Transition matrix is valid.")
else:
    print("✗ Transition matrix is invalid.")


# ------------------------------------------------------------
# 5. CREATE GOOGLE MATRIX
# ------------------------------------------------------------

# Equal teleportation probability
v = np.ones(n) / n

# Google Matrix:
#
# G = dP + (1-d)v1^T

G = damping * P + (1 - damping) * np.outer(v, np.ones(n))

print("\n" + "=" * 65)
print("GOOGLE MATRIX G")
print("=" * 65)

G_table = pd.DataFrame(
    G,
    index=[f"Page {p}" for p in pages],
    columns=[f"Page {p}" for p in pages]
)

display(G_table.style.format("{:.6f}"))


# ------------------------------------------------------------
# 6. POWER ITERATION
# ------------------------------------------------------------

# Initially all pages have equal importance
rank = np.ones(n) / n

history = [rank.copy()]

for iteration in range(1, max_iterations + 1):

    # Power method:
    #
    # r(k+1) = G r(k)

    new_rank = G @ rank

    # Calculate convergence error
    error = np.linalg.norm(
        new_rank - rank,
        ord=1
    )

    rank = new_rank

    history.append(rank.copy())

    # Stop when convergence is reached
    if error < tolerance:
        break


history = np.array(history)


# ------------------------------------------------------------
# 7. FINAL PAGERANK RESULTS
# ------------------------------------------------------------

results = pd.DataFrame({
    "Page": pages,
    "PageRank": rank,
    "Percentage": rank * 100
})

results = results.sort_values(
    by="PageRank",
    ascending=False
).reset_index(drop=True)


print("\n" + "=" * 65)
print("FINAL PAGERANK RESULTS")
print("=" * 65)

display(
    results.style.format({
        "PageRank": "{:.6f}",
        "Percentage": "{:.2f}%"
    })
)

most_important = results.iloc[0]["Page"]

print("\n⭐ MOST IMPORTANT PAGE")
print("---------------------")
print(f"Page {most_important}")
print(f"PageRank = {results.iloc[0]['PageRank']:.6f}")


# ------------------------------------------------------------
# 8. EIGENVALUE = 1 VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("EIGENVALUE = 1 VERIFICATION")
print("=" * 65)

# At steady state:
#
# G r = r
#
# Therefore:
#
# G r = 1r
#
# So r is an eigenvector with eigenvalue 1.

Gr = G @ rank

verification_error = np.linalg.norm(
    Gr - rank
)

print("G × r:")
print(Gr)

print("\nr:")
print(rank)

print("\n||G r - r||:")
print(verification_error)

if verification_error < tolerance:
    print("\n✓ PageRank is a steady-state eigenvector.")
    print("✓ Eigenvalue λ ≈ 1")


# ------------------------------------------------------------
# 9. DIRECT EIGENVALUE CALCULATION
# ------------------------------------------------------------

eigenvalues, eigenvectors = np.linalg.eig(G)

index = np.argmin(
    np.abs(eigenvalues - 1)
)

closest_eigenvalue = np.real(
    eigenvalues[index]
)

raw_vector = np.real(
    eigenvectors[:, index]
)

# Eigenvectors may have a negative sign.
if np.sum(raw_vector) < 0:
    raw_vector = -raw_vector

# Normalize eigenvector
eigen_rank = raw_vector / np.sum(raw_vector)

print("\n" + "=" * 65)
print("DIRECT EIGENVALUE CALCULATION")
print("=" * 65)

print("\nEigenvalues:")

for value in eigenvalues:
    print(f"{value}")

print("\nEigenvalue closest to 1:")
print(f"{closest_eigenvalue:.12f}")

print("\nEigenvector corresponding to λ ≈ 1:")

for page, value in zip(pages, eigen_rank):
    print(f"Page {page}: {value:.6f}")


# ------------------------------------------------------------
# 10. COMPARE POWER METHOD AND EIGENVECTOR
# ------------------------------------------------------------

difference = np.linalg.norm(
    eigen_rank - rank
)

print("\nDifference between methods:")
print(f"{difference:.12e}")

if difference < 1e-8:
    print("✓ Both methods give the same PageRank.")
else:
    print("⚠ Small numerical difference detected.")


# ------------------------------------------------------------
# 11. CHART 1 — POWER ITERATION CONVERGENCE
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

for i, page in enumerate(pages):

    plt.plot(
        range(len(history)),
        history[:, i],
        marker="o",
        markersize=3,
        label=f"Page {page}"
    )

plt.title(
    "PageRank Convergence Using Power Iteration",
    fontsize=15
)

plt.xlabel("Iteration")
plt.ylabel("PageRank")

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. CHART 2 — FINAL PAGERANK
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

bars = plt.bar(
    results["Page"],
    results["PageRank"]
)

plt.title(
    "Final PageRank of the 5 Web Pages",
    fontsize=15
)

plt.xlabel("Web Page")
plt.ylabel("PageRank")

plt.grid(
    axis="y",
    alpha=0.3
)

for bar, value in zip(
    bars,
    results["PageRank"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.3f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 13. CHART 3 — PAGERANK DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 7))

plt.pie(
    results["PageRank"],
    labels=[
        f"Page {p}"
        for p in results["Page"]
    ],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("PageRank Distribution")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 14. WEB LINK VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(9, 7))

positions = {
    "A": (0, 1),
    "B": (1, 2),
    "C": (2, 1),
    "D": (1, 0),
    "E": (3, 1)
}

# Draw pages
for page, (x, y) in positions.items():

    plt.scatter(
        x,
        y,
        s=1800,
        alpha=0.7
    )

    plt.text(
        x,
        y,
        f"Page {page}",
        ha="center",
        va="center",
        fontweight="bold"
    )


# Draw arrows
for source, destinations in links.items():

    x1, y1 = positions[source]

    for destination in destinations:

        x2, y2 = positions[destination]

        plt.annotate(
            "",
            xy=(x2, y2),
            xytext=(x1, y1),
            arrowprops=dict(
                arrowstyle="->",
                lw=1.5
            )
        )

plt.title(
    "Link Structure of the Mini Web",
    fontsize=15
)

plt.axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. BENCHMARK TEST
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("BENCHMARK TEST")
print("=" * 65)

# Reuse the same realistic 5-page dataset
benchmark_P = P.copy()

benchmark_rank = np.ones(n) / n

for benchmark_iteration in range(
    1,
    max_iterations + 1
):

    new_benchmark_rank = (
        G @ benchmark_rank
    )

    benchmark_error = np.linalg.norm(
        new_benchmark_rank - benchmark_rank,
        ord=1
    )

    benchmark_rank = new_benchmark_rank

    if benchmark_error < tolerance:
        break


print("Benchmark iterations:",
      benchmark_iteration)

print(
    "Benchmark final error:",
    f"{benchmark_error:.12e}"
)

print(
    "Rank sum:",
    f"{benchmark_rank.sum():.12f}"
)


# Automatic tests
test_1 = np.isclose(
    benchmark_rank.sum(),
    1
)

test_2 = np.all(
    benchmark_rank >= 0
)

test_3 = (
    benchmark_error < tolerance
)

test_4 = np.allclose(
    benchmark_rank,
    rank,
    atol=1e-8
)


print("\nBenchmark checks:")
print("✓ Rank sums to 1:", test_1)
print("✓ All ranks are non-negative:", test_2)
print("✓ Power iteration converged:", test_3)
print("✓ Benchmark matches main result:", test_4)


if test_1 and test_2 and test_3 and test_4:
    print("\n✅ BENCHMARK TEST PASSED")
else:
    print("\n❌ BENCHMARK TEST FAILED")


# ------------------------------------------------------------
# 16. FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("FINAL PROJECT SUMMARY")
print("=" * 65)

print("""
The PageRank algorithm models a web as a network of pages.

The transition matrix P represents the probability of moving
from one page to another.

The Google matrix G includes both link-following and
random teleportation.

Power iteration repeatedly calculates:

        r(k+1) = G r(k)

When the process reaches a steady state:

        G r = r

which is equivalent to:

        G r = 1r

Therefore, the PageRank vector is an eigenvector
corresponding to eigenvalue λ = 1.
""")

print("FINAL RANKING")
print("-" * 40)

for i, row in results.iterrows():

    print(
        f"{i + 1}. Page {row['Page']}  "
        f"→ {row['PageRank']:.6f}"
    )

print("\n⭐ MOST IMPORTANT PAGE:",
      f"Page {most_important}")

print(
    f"⭐ PageRank = "
    f"{results.iloc[0]['PageRank']:.6f}"
)

print("\n" + "=" * 65)
print("             PROJECT COMPLETED SUCCESSFULLY")
print("=" * 65)
```
